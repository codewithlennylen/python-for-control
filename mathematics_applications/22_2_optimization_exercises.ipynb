## 22.0.1 Make your own minimization function



- Implement a basic minimization function from scratch using either a For or While Loop
- Test the function and get the answer (-1)

### Setup


```python
import numpy as np
import matplotlib.pyplot as plt


x_min = -5
x_max = 5
dx = 0.1

N = int((x_max - x_min)/dx)

x = np.arange(x_min, x_max+dx, dx)

def y(x):
    return x**2 + 2*x + 1



plt.plot(x, y(x))
```

### 22.0.1 Solution

```python
def minimize_f(function, x):
    y_min = float('inf')
    x_min = float('inf')

    for i in x:
        y = function(i)
        if y < y_min:
            y_min = y
            x_min = i

    return int(x_min)


print(minimize_f(y, x))
```
