## 22.0.1 Make your own minimization function



- Implement a basic minimization function from scratch using either a For or While Loop
- Test the function and get the answer (-1)

### Setup


```python
import numpy as np
import matplotlib.pyplot as plt


x_min = -5
x_max = 5
dx = 0.1

# N = int((x_max - x_min)/dx)

x = np.arange(x_min, x_max+dx, dx)
print(x)

def y(x):
    return x**2 + 2*x + 1

# for i in x:
#     print(i, y(i))


plt.plot(x, y(x))
```

### 22.0.1 Solution

```python
def minimize_f(function, x):
    y_min = float('inf')
    x_min = float('inf')

    for i in x:
        y = function(i)
        if y < y_min:
            y_min = y
            x_min = i

    return int(x_min)


print(minimize_f(y, x))
```

## 22.0.2 Compare Python Optimization Functions



```python
from scipy import optimize

x_minimum_fmin = optimize.fmin(y, x0=0)
x_minimum_scalar = optimize.minimize_scalar(y, (x_min, x_max)) 
x_minimum_minimize = optimize.minimize(y ,x0=0) # Objective function must return a scalar
x_minimum_bound = optimize.fminbound(y, x_min, x_max)

print(f'scipy.optimize.fmin: {x_minimum_fmin}')
print(f'scipy.optimize.minimize_scalar: {x_minimum_scalar.x}')
print(f'scipy.optimize.minimize: {x_minimum_minimize.x}')
print(f'scipy.optimize.fminbound: {x_minimum_bound}')
```
