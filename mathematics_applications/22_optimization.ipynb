# Optimization

Optimization is based on finding the minimum of a given criteria function.



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import optimize


x_min = -5
x_max = 5
dx = 0.1

N = int((x_max-x_min)/dx) # -> resolves to 100

x = np.arange(x_min, x_max+dx, dx)

y = x**2 + 2*x + 1

def y_(x):
    return x**2 + 2*x + 1


x_minimum = optimize.fminbound(y_, x_min, x_max)

print(f'x_minimum: {x_minimum}')

plt.plot(x, y)
plt.xlim([x_min, x_max])

```
